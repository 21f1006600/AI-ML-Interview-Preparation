# Loading the dataset

```python
BASE_PATH = "/kaggle/input/chest-xray-pneumonia/chest_xray/chest_xray"

TRAIN_PATH = BASE_PATH + "/train"
VAL_PATH   = BASE_PATH + "/val"
TEST_PATH  = BASE_PATH + "/test"

print("Train:", TRAIN_PATH)
print("Validation:", VAL_PATH)
print("Test:", TEST_PATH)
```

```python
import torch
import torchvision

print("Pytorch version:", torch.__version__)
print("Torchvision version", torchvision.__version__)
print("GPU available:", torch.cuda.is_available())
```

# Loading one X-Ray Image

```python
from PIL import Image
import os

#get the normal folder
normal_folder = os.path.join(TRAIN_PATH, "NORMAL")

#get the first image filename
image_name = os.listdir(normal_folder)[0]

#create the complete path
image_path = os.path.join(normal_folder, image_name)

#open the image
image = Image.open(image_path)

print("Image name:", image_name)
print("Image type", type(image))
print("Image size", image.size)
print("Image mode", image.mode)
```

#Displaying the X-Ray

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.imshow(image, cmap='gray')
plt.axis('off')
plt.title("Normal Chest X-Ray")
plt.show()
```

#Converting the X-Ray into a Pytorch tensor

```python
import torch
from torchvision import transforms

#converting the PIL image into pytorch tensor
transform = transforms.ToTensor()

image_tensor = transform(image)

print("Tensor shape:", image_tensor.shape)
print("Tensor data type:", image_tensor.dtype)
print("Minimum value:", image_tensor.min())
print("Maximum value:", image_tensor.max())
```

#Resize the image

```python
resize_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

resized_tensor = resize_transform(image)

print('Original size',image.size)
print('Tensor shape after resize', resized_tensor.shape)
```

#Creating the PyTorch Dataset

```python
from torchvision import datasets

train_dataset = datasets.ImageFolder(
    root=TRAIN_PATH,
    transform=resize_transform
)
```
